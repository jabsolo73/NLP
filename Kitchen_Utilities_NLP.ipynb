```python
import pandas as pd
from textblob import TextBlob

# Load the data from the Excel file
# Replace 'path_to_excel_file.xlsx' with the actual path to your Excel file
df = pd.read_excel('/content/Column_taken_file.xlsx')

# Ensure that the 'host_about' column is treated as a string
df['host_about'] = df['host_about'].astype(str)

# Define a function to analyze sentiment
def analyze_sentiment(review):
    # Convert NaN values to string 'nan' and then to an empty string
    if pd.isna(review):
        review = ''
    analysis = TextBlob(review)
    return analysis.sentiment.polarity, analysis.sentiment.subjectivity

# Apply the function to each review in the 'host_about' column
df['Polarity'], df['Subjectivity'] = zip(*df['host_about'].map(analyze_sentiment))

# Output the DataFrame to a new Excel file
df.to_excel('sentiment_analysis_output.xlsx', index=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming df is your DataFrame with the Polarity and Subjectivity columns
# Categorize sentiments
def categorize_sentiment(polarity):
    if polarity > 0.1:
        return 'Positive'
    elif polarity < -0.1:
        return 'Negative'
    else:
        return 'Neutral'

df['Sentiment'] = df['Polarity'].apply(categorize_sentiment)

# Visualize the distribution of polarity scores
plt.figure(figsize=(8, 6))
df['Polarity'].hist(bins=20)
plt.title('Distribution of Polarity Scores')
plt.xlabel('Polarity Score')
plt.ylabel('Frequency')
plt.show()

# Visualize the distribution of sentiment categories
sentiment_counts = df['Sentiment'].value_counts(normalize=True) * 100
sentiment_counts.plot(kind='bar', color=['green', 'blue', 'red'])
plt.title('Distribution of Sentiment Categories')
plt.xlabel('Sentiment')
plt.ylabel('Percentage')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud

# Load your DataFrame here
df = pd.read_excel('/content/sentiment_analysis_output.xlsx')

# Convert all non-string data to strings and handle NaN values
df['amenities'] = df['amenities'].fillna('').astype(str)

# Concatenate all amenities into a single string
all_amenities = ' '.join(df['amenities'])

# Create the word cloud object
wordcloud = WordCloud(width=800, height=400, background_color='white').generate(all_amenities)

# Display the word cloud using matplotlib
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')  # Hide the axes
plt.show()
```

```python
# Split the amenities string by the delimiter and count the number of elements
df['Amenities_Count'] = df['amenities'].str.split(',').apply(lambda x: len(x) if isinstance(x, list) else 0)
df.to_excel('new_file.xlsx', index=False)
```
